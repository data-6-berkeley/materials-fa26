# Control: Conditional statements

```python
x = -1
#Determine the polarity of the numeric input
if x > 0:
    print('Positive')

elif x < 0:
    print('Negative')

elif x == 0:
    print('Neither positive nor negative')
```

```python
x = 0
#Determine the polarity of the numeric input
if x > 0:
    print('Positive')

elif x < 0:
    print('Negative')

elif x == 0:
    print('Neither positive nor negative')
```

```python
x = "-1"
#Determine the polarity of the numeric input
if x > 0:
    print('Positive')

elif x < 0:
    print('Negative')

elif x == 0:
    print('Neither positive nor negative')
```
