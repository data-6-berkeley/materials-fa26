## Calculating Summary Statistics

```python
#3 ways to find mean
values = make_array(2, 3, 3, 9)

np.sum(values)/len(values)
np.average(values)
np.mean(values)
```

**Standard Deviation** (SD): root mean squared deviation from average = square root of the variance

## Spread: Standard deviation also helps us quantify “variability”

For this class, we do not focus much on standard deviation. We'd prefer InterQuartile Range where possible. Take Data 8 to learn more about standard deviations.

```python
values = make_array(2, 3, 3, 9)
sd_table = Table().with_columns('Value', values)
sd_table
```

```python
average_value = np.mean(values)
average_value
```

```python
deviations = values - average_value
deviations
```

```python
sum(deviations)
```

```python
sd_table = sd_table.with_column('Deviation', deviations)
sd_table
```

**Variance** of the data -  mean squared deviation from average

```python
variance = np.mean(deviations ** 2)
variance
```

```python
sd = variance ** 0.5
sd
```
