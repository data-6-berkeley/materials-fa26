# While-loops

## Repeat while the condition is *true*

```python
counter = 10
while counter > 0:
    print(counter)
    counter -= 1
print("blast off!")
```

### Discussion: What is the difference between these two loops?

Every for-loop can be written as a while-loop, but not vice versa!

```python
values = make_array(3, 4, 5)
total = 0
for v in values:
    total += v ** 2
print(total)
```

```python
values = make_array(3, 4, 5)
total = 0
i = 0
while i < len(values):
    total += values.item(i) ** 2
    i += 1
print(total)
```
